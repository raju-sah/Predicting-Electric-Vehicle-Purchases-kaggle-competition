```python
# ── Core Libraries & PyTorch Setup ──
import os
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import roc_auc_score, roc_curve

# Hardware verification
device = torch.device('cpu')
print(f"🚀 PyTorch Hardware Backend: {device}")

# TokyoNight Dark Styling
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'
plt.rcParams['axes.edgecolor'] = '#414868'
plt.rcParams['axes.linewidth'] = 1.2

COLOR_CYAN = '#7aa2f7'
COLOR_PURPLE = '#bb9af7'
COLOR_ORANGE = '#ff9e64'
COLOR_GREEN = '#9ece6a'
COLOR_RED = '#f7768e'
BG_SURFACE = '#1a1b26'
print("✓ PyTorch tabular engine initialized.")
```

## 📥 2. Tabular Ingestion & Embedding Preparation
We clean and partition features into continuous numerical columns and categorical columns for embedding lookup.

```python
comp_dir = '/kaggle/input/playground-series-s6e9'
if not os.path.exists(comp_dir):
    matches = glob.glob('/kaggle/input/**/train.csv', recursive=True)
    comp_dir = os.path.dirname(matches[0]) if matches else '.'

train_df = pd.read_csv(os.path.join(comp_dir, 'train.csv'))
test_df = pd.read_csv(os.path.join(comp_dir, 'test.csv'))

if train_df['Will_Buy_EV'].dtype == object:
    y = (train_df['Will_Buy_EV'].str.strip().str.lower() == 'yes').astype(int).values
else:
    y = train_df['Will_Buy_EV'].astype(int).values

# Feature Engineering Ratios
for d in [train_df, test_df]:
    home_cs = d['Charging_Stations_Near_Home'].fillna(0)
    work_cs = d['Charging_Stations_Near_Work'].fillna(0)
    d['Total_Charging'] = home_cs + work_cs
    d['Charging_Ratio'] = (home_cs + 0.1) / (work_cs + home_cs + 0.2)
    cars = d['Number_of_Cars_Owned'].fillna(1).clip(lower=1)
    d['Income_Per_Car'] = d['Annual_Income_USD'] / cars
    d['Log_Income'] = np.log1p(d['Annual_Income_USD'].clip(lower=0))

cat_cols = ['Gender', 'City_Type', 'Current_Car_Type', 'Home_Charging_Possible', 'Subsidy_Available', 'Range_Anxiety_Level']
cont_cols = ['Age', 'Annual_Income_USD', 'Log_Income', 'Daily_Commute_km', 'Number_of_Cars_Owned', 'Income_Per_Car',
             'Charging_Stations_Near_Home', 'Charging_Stations_Near_Work', 'Total_Charging', 'Charging_Ratio', 'Environmental_Concern_Level']

# Clean Non-Negative Label Encoding for Embeddings (Zero Leakage)
cat_dims = []
for col in cat_cols:
    train_df[col] = train_df[col].fillna('Missing').astype(str)
    test_df[col] = test_df[col].fillna('Missing').astype(str)
    unique_vals = sorted(list(set(train_df[col].unique()) | set(test_df[col].unique())))
    val2idx = {val: i for i, val in enumerate(unique_vals)}
    train_df[col] = train_df[col].map(val2idx).fillna(0).astype(int)
    test_df[col] = test_df[col].map(val2idx).fillna(0).astype(int)
    cat_dims.append(len(unique_vals))

# Embedding Dimensions
emb_dims = [(num_c, min(16, max(4, (num_c + 1) // 2))) for num_c in cat_dims]
print(f"📐 Categorical Embedding Dimensions: {list(zip(cat_cols, emb_dims))}")

# Standardize Continuous Features
scaler = StandardScaler()
X_cont = scaler.fit_transform(train_df[cont_cols].fillna(0))
X_test_cont = scaler.transform(test_df[cont_cols].fillna(0))

X_cat = train_df[cat_cols].values
X_test_cat = test_df[cat_cols].values
print(f"✓ Data split: {X_cont.shape[1]} continuous features, {X_cat.shape[1]} categorical features.")
```

## 🏗️ 3. PyTorch Tabular ResNet Architecture
We implement a modular **Tabular ResNet** combining embedding tables with dense residual blocks.

```python
class TabularDataset(Dataset):
    def __init__(self, cont_data, cat_data, labels=None):
        self.cont = torch.tensor(cont_data, dtype=torch.float32)
        self.cat = torch.tensor(cat_data, dtype=torch.long)
        self.labels = torch.tensor(labels, dtype=torch.float32) if labels is not None else None

    def __len__(self):
        return len(self.cont)

    def __getitem__(self, idx):
        if self.labels is not None:
            return self.cont[idx], self.cat[idx], self.labels[idx]
        return self.cont[idx], self.cat[idx]

class ResidualBlock(nn.Module):
    def __init__(self, hidden_dim, dropout=0.15):
        super().__init__()
        self.fc1 = nn.Linear(hidden_dim, hidden_dim)
        self.bn1 = nn.BatchNorm1d(hidden_dim)
        self.act1 = nn.SiLU()
        self.drop = nn.Dropout(dropout)
        self.fc2 = nn.Linear(hidden_dim, hidden_dim)
        self.bn2 = nn.BatchNorm1d(hidden_dim)
        self.act2 = nn.SiLU()

    def forward(self, x):
        residual = x
        out = self.drop(self.act1(self.bn1(self.fc1(x))))
        out = self.bn2(self.fc2(out))
        return self.act2(out + residual)

class TabularResNet(nn.Module):
    def __init__(self, emb_dims, num_cont, hidden_dim=128):
        super().__init__()
        self.embeddings = nn.ModuleList([nn.Embedding(num_c, emb_d) for num_c, emb_d in emb_dims])
        total_emb_dim = sum(emb_d for _, emb_d in emb_dims)
        
        in_dim = total_emb_dim + num_cont
        self.input_layer = nn.Sequential(
            nn.Linear(in_dim, hidden_dim),
            nn.BatchNorm1d(hidden_dim),
            nn.SiLU()
        )
        
        self.res1 = ResidualBlock(hidden_dim, dropout=0.15)
        self.res2 = ResidualBlock(hidden_dim, dropout=0.15)
        self.head = nn.Linear(hidden_dim, 1)

    def forward(self, x_cont, x_cat):
        emb_list = [emb(x_cat[:, i]) for i, emb in enumerate(self.embeddings)]
        x_emb = torch.cat(emb_list, dim=1)
        x = torch.cat([x_cont, x_emb], dim=1)
        
        h = self.input_layer(x)
        h = self.res1(h)
        h = self.res2(h)
        logits = self.head(h).squeeze(-1)
        return logits

print("✓ TabularResNet architecture ready.")
```

## 🚀 4. Stratified 5-Fold Deep Learning Training Loop
We train across 5 stratified folds with batch size $2048$ using AdamW and Cosine Annealing.

```python
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
oof_neural = np.zeros(len(y))
test_neural = np.zeros(len(X_test_cont))

test_dataset = TabularDataset(X_test_cont, X_test_cat)
test_loader = DataLoader(test_dataset, batch_size=4096, shuffle=False)

BATCH_SIZE = 2048
EPOCHS = 5
criterion = nn.BCEWithLogitsLoss()

print("Training PyTorch Tabular ResNet across 5 Folds...")

for fold, (trn_idx, val_idx) in enumerate(skf.split(X_cont, y)):
    trn_ds = TabularDataset(X_cont[trn_idx], X_cat[trn_idx], y[trn_idx])
    val_ds = TabularDataset(X_cont[val_idx], X_cat[val_idx], y[val_idx])
    trn_loader = DataLoader(trn_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
    val_loader = DataLoader(val_ds, batch_size=4096, shuffle=False)
    
    model = TabularResNet(emb_dims, X_cont.shape[1], hidden_dim=128).to(device)
    optimizer = optim.AdamW(model.parameters(), lr=3e-3, weight_decay=1e-4)
    scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)
    
    for epoch in range(EPOCHS):
        model.train()
        for c_cont, c_cat, lbl in trn_loader:
            c_cont, c_cat, lbl = c_cont.to(device), c_cat.to(device), lbl.to(device)
            optimizer.zero_grad()
            logits = model(c_cont, c_cat)
            loss = criterion(logits, lbl)
            loss.backward()
            optimizer.step()
        scheduler.step()
        
    # Validation Evaluation
    model.eval()
    val_preds = []
    with torch.no_grad():
        for c_cont, c_cat, _ in val_loader:
            c_cont, c_cat = c_cont.to(device), c_cat.to(device)
            probs = torch.sigmoid(model(c_cont, c_cat)).cpu().numpy()
            val_preds.extend(probs)
            
    val_preds = np.array(val_preds)
    oof_neural[val_idx] = val_preds
    fold_auc = roc_auc_score(y[val_idx], val_preds)
    print(f"  • Fold {fold+1} Neural ROC-AUC: {fold_auc:.5f}")
    
    # Test Inference
    test_preds_fold = []
    with torch.no_grad():
        for c_cont, c_cat in test_loader:
            c_cont, c_cat = c_cont.to(device), c_cat.to(device)
            probs = torch.sigmoid(model(c_cont, c_cat)).cpu().numpy()
            test_preds_fold.extend(probs)
    test_neural += np.array(test_preds_fold) / 5.0

overall_neural_auc = roc_auc_score(y, oof_neural)
print(f"\n🏆 PyTorch Tabular ResNet Out-Of-Fold ROC-AUC: {overall_neural_auc:.5f}")
```

## 📈 5. Neural Network Diagnostic ROC Curve
We inspect the Receiver Operating Characteristic curve for the PyTorch Tabular ResNet.

```python
fig, ax = plt.subplots(figsize=(9, 5.5), facecolor=BG_SURFACE)
ax.set_facecolor(BG_SURFACE)

fpr, tpr, _ = roc_curve(y, oof_neural)
ax.plot(fpr, tpr, color=COLOR_PURPLE, linewidth=2.5, label=f'PyTorch Tabular ResNet (AUC = {overall_neural_auc:.4f})')
ax.plot([0, 1], [0, 1], color=COLOR_ORANGE, linestyle='--', label='Random Chance (0.5000)')

ax.set_title("PyTorch Tabular ResNet ROC Curve (5-Fold Cross-Validation)", color='white', fontsize=13, pad=15)
ax.set_xlabel("False Positive Rate (1 - Specificity)", color='#c0caf5', fontsize=11)
ax.set_ylabel("True Positive Rate (Sensitivity)", color='#c0caf5', fontsize=11)
ax.grid(True, alpha=0.15)
ax.legend(facecolor='#24283b', edgecolor='#414868', labelcolor='white')

plt.tight_layout()
plt.show()
```

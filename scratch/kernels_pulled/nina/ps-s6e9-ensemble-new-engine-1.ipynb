```python
# ====================================================================
# PRODUCTION ENGINE: Clean 3-Way Rank-Average Ensemble Pipeline
# Optimized with Professional Spearman Correlation & KDE Diagnostics
# Co-developed by Nina & Google AI Asset Management Group
# Target: Ultimate Calibration Check Beyond the 0.94643 Barrier
# ====================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')

# --------------------------------------------------------------------
# STEP 1: Loading and Renaming Target Submissions from Dataset v9
# --------------------------------------------------------------------
print("Loading strategic pilots into the analytical core...")
path = '/kaggle/input/datasets/nina2025/ps-s6-e9-09/'

# Reading the exact files from your repository
df_models = pd.DataFrame()
df_models['id'] = pd.read_csv('/kaggle/input/competitions/playground-series-s6e9/sample_submission.csv')['id']

df_models['Model_43_OOF'] = pd.read_csv(path + '0.94643 - ver.6 (OOF Power-Two single models Blend).csv')['Will_Buy_EV']
df_models['Model_42_v4'] = pd.read_csv(path + '0.94642 - version 4 (h-blend(3)).csv')['Will_Buy_EV']
df_models['Model_39_v18'] = pd.read_csv(path + '0.94639 - version_18 (h-blend high speed).csv')['Will_Buy_EV']

# --------------------------------------------------------------------
# STEP 2: Transforming into Perfect 0-1 Percentile Ranks
# --------------------------------------------------------------------
print("Transforming probability vectors into perfect percentile ranks...")
df_ranks = pd.DataFrame()
df_ranks['Model_43_OOF'] = df_models['Model_43_OOF'].rank(pct=True)
df_ranks['Model_42_v4'] = df_models['Model_42_v4'].rank(pct=True)
df_ranks['Model_39_v18'] = df_models['Model_39_v18'].rank(pct=True)

# --------------------------------------------------------------------
# STEP 3: Computing Weighted Rank-Average with Your Exact Mass Split
# --------------------------------------------------------------------
print("Applying tactical weight matrix [70% + 20% + 10%]...")
df_models['Final_Blend'] = (df_ranks['Model_43_OOF'] * 0.70 + 
                             df_ranks['Model_42_v4'] * 0.20 + 
                             df_ranks['Model_39_v18'] * 0.10).rank(pct=True)

# Save the final clear submission to disk
submission = df_models[['id', 'Final_Blend']].rename(columns={'Final_Blend': 'Will_Buy_EV'})
submission.to_csv('submission.csv', index=False)
print("The master champion submission.csv successfully saved to disk!")

# --------------------------------------------------------------------
# STEP 4: PROFESSIONAL GRAPHICAL DIAGNOSTICS (Spearman & KDE)
# --------------------------------------------------------------------
print("\nGenerating professional diagnostic plots...")

plt.style.use('seaborn-v0_8-whitegrid')
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Plot 1: Spearman Rank Correlation Heatmap
# We use Spearman correlation because ROC-AUC depends purely on rank order
corr_matrix = df_models[['Model_43_OOF', 'Model_42_v4', 'Model_39_v18']].corr(method='spearman')
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

sns.heatmap(corr_matrix, mask=mask, annot=True, cmap='RdBu_r', center=1.0,
            fmt='.4f', square=True, linewidths=0.5, ax=axes[0], cbar_kws={'shrink': 0.8})
axes[0].set_title('Spearman Rank Correlation Matrix', fontweight='bold', fontsize=11)
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation=30, ha='right')

# Plot 2: Kernel Density Estimation (KDE) Scale Check
# Verifying that the final blend maintains a perfect uniform distribution for ROC-AUC
sns.kdeplot(df_models['Model_43_OOF'], label='Model 43 OOF (70%)', color='navy', linewidth=1.0, ax=axes[1])
sns.kdeplot(df_models['Model_42_v4'], label='Model 42 v4 (20%)', color='darkorange', linewidth=1.0, ax=axes[1])
sns.kdeplot(df_models['Model_39_v18'], label='Model 39 v18 (10%)', color='darkgreen', linewidth=1.0, ax=axes[1])
sns.kdeplot(df_models['Final_Blend'], label='Final Rank Blend', color='crimson', linewidth=2.0, linestyle='--', ax=axes[1])

axes[1].set_title('Kernel Density Estimation (KDE) Distribution', fontweight='bold', fontsize=11)
axes[1].set_xlabel('Percentile Rank / Probability Scale')
axes[1].set_ylabel('Density Frequency')
axes[1].legend(loc='upper right', fontsize=9)

plt.tight_layout()
plt.show()
```
